```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os
```

```python
# Folder containing your 12 CSV files
folder = r"D:\Projects\Cyclist Capstone Project\Dataset"
```

```python
# Find all CSV files
files = glob.glob(os.path.join(folder, "*.csv"))

print("Files found:", len(files))
```

```python
# Read and combine all files
df_list = []

for file in files:
    print("Reading:", os.path.basename(file))
    df = pd.read_csv(file)
    df_list.append(df)
```

```python
# Combine all monthly files
combined_df = pd.concat(df_list, ignore_index=True)

# Show basic information
print("\nTotal rows:", len(combined_df))
print("Total columns:", len(combined_df.columns))
print("\nColumns:")
print(combined_df.columns.tolist())
```

```python
# Save combined dataset
output_file = os.path.join(folder, "cyclistic_combined.csv")
combined_df.to_csv(output_file, index=False)

print("\nCombined file saved to:")
print(output_file)
```
